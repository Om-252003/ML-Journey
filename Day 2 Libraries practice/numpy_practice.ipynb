```python
import numpy as np
```

```python
from time import process_time
```

```python
l1 = [i for i in range(1000000)]

start = process_time()

l1 = [i+5 for i in l1]

end = process_time()

print(end-start)
```

```python
numpyArray = np.array([i for i in range(1000000)])
start = process_time()

numpyArray += 5 #It will auto do the addition of every element

end = process_time()

print(end-start)
```

```python
o = np.random.randint(1,5,(3,3))
p = np.random.randint(6,10,(3,3))

print("o \n",o)
print("p \n",p)
print("o+p \n",o+p)
print("o-p \n",o-p)
print("o*p \n",o*p)
print("p//o \n",p//o)
print("o%p \n",o%p)
```
