Matplotlib

* For making Plots

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)
z = np.cos(x)

print(x)
```

```python
%matplotlib inline

# Sine Wave
plt.plot(x,y)
plt.show()
```

```python
%matplotlib inline

# Cosine Wave
plt.plot(x,z)

plt.xlabel('angle')
plt.ylabel('cos value')

plt.title('Cosine Wave')


plt.show()
```

```python
%matplotlib inline
# Parabola

x = np.linspace(-10,10,20)
y = x**2

plt.plot(x,y)
plt.show()
```

```python
%matplotlib inline

plt.plot(x,y, 'r+')
plt.show()
```

```python
%matplotlib inline

plt.plot(x,y,'gx')
plt.show()
```

```python
%matplotlib inline

x = np.linspace(-5,5,50)
plt.plot(x, np.sin(x), 'g')
plt.plot(x, np.cos(x), 'r--')
plt.show()
```

```python
%matplotlib inline

# Barplot

fig = plt.figure()

ax = fig.add_axes([0,0,1,1])

langs = ['Python', 'Java', 'C++', 'JavaScript', 'C']
people = [80, 60, 40, 70, 30]

ax.bar(langs,people) # This creates the bar plot

plt.xlabel("Languages")
plt.ylabel("People")
plt.title("Barplot")

plt.show()
```

```python
%matplotlib inline

# Pie Chart
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])

langs = ['Python', 'Java', 'C++', 'JavaScript', 'C']
people = [80, 60, 40, 70, 30]

ax.pie(people, labels=langs, autopct='%1.2f%%') # this creates the pie chart

plt.show()

# autopct='%1.2f%%' this meanss give only 2 numbers after decimal point 
```

```python
%matplotlib inline

# Scatter Plot
x = np.linspace(0,10,50)
y = np.sin(x)
z = np.cos(x)

fig = plt.figure()
ax = fig.add_axes([0,0,1,1])

ax.scatter(x,y,color= 'g') # this created scatter plot
ax.scatter(x,z,color= 'b')

plt.show()
```

```python
%matplotlib inline

# 3D Scatter Plot

fig = plt.figure()
ax = plt.axes(projection = '3d')

x = 20 * np.random.random(100)
y = np.sin(x)
z = np.cos(x)

ax.scatter(y,z,x, c=x, cmap='Blues')
plt.show()
```
