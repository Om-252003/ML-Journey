```python
random_int_matrix = np.random.randint(100, size= (4,5))
print(random_int_matrix)
```

```python
mat_of_1 = np.ones((4,5))  #floating decimal values
print(mat_of_1)
```

```python
mat_of_1 = np.ones((4,5), dtype=int)
print(mat_of_1)
```

```python
mat_of_0 = np.zeros((4,5), dtype=int)
print(mat_of_0)
```

```python
identity_matrix = np.eye(5, dtype=int)
print(identity_matrix)
```

```python
transpose_of_matrix = np.transpose(random_int_matrix)
print(transpose_of_matrix)
```

```python
# matric * matrix

m1 = np.array([[2,3], [10,5]])
m2 = np.array([[10,5], [20,4]])

# dot operation
m3 = m1 @ m2        # it is np.dot(m1,m2)
print(m3)

# element wise multiplication
m3 = np.multiply(m1,m2)
print(m3)

```
