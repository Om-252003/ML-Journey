```python
# creating vectors


v1 = np.asarray([0,0,2,3])
v2 = np.asarray([0,0,3,-2])
```

```python
plt.quiver(v1[0], v1[1], v1[2], v1[3], scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(v2[0], v2[1], v2[2], v2[3], scale_units='xy', angles='xy', scale=1, color='y')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

```python
 # Adding vectors

v3 = v1 + v2
print(v3)
```

```python
# '*' unzips the list and passes the elements as single single values

plt.quiver(*v2, scale_units='xy', angles='xy', scale=1, color='y')
plt.quiver(*v1, scale_units='xy', angles='xy', scale=1, color='b')
plt.quiver(*v3, scale_units='xy', angles='xy', scale=1, color='g')
plt.xlim(-8,8)
plt.ylim(-8,8)
plt.show()
```

```python
# subtraction

v4 = v1-v2
print(v4)
```

```python
# Dot product of 2 vectors (results a scalar)

a = np.asarray([2,3])
b = np.asarray([4,4])

a_dot_b = np.dot(a,b)
print(a_dot_b)
```

```python
# Cross Product of 2 vectors  (results a vector) (similar to matrix multiplication)

a = np.asarray([2,3])
b = np.asarray([4,4])

a_cross_b = np.cross(a,b)
print(a_cross_b)

print(type(a_cross_b))
```

```python
# projection of 'a' vector on 'v vector (results a vector)

a = np.asarray([2,5])
v = np.asarray([8,-6])

# magnitude of v vector (square root of sum of square of all values)
magnitude_of_v = np.sqrt(sum(v**2))  # sqrt(sum(8**2, -6**2))
print("magnitude_of_v :",magnitude_of_v)

#projecion of a on v
projection_of_a_on_v = (np.dot(a,v)/magnitude_of_v ** 2) *v
print("projection_of_a_on_v :",projection_of_a_on_v)
```
