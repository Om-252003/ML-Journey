# 1) Imputation
**Central Tendencies:**
  1. Mean (Average Value)
  2. Median (Central Value)
  3. Mode (Most common value)

```python
# Analyse the data distribution

fig , ax = plt.subplots(figsize=(8,8))
sns.distplot(df.salary)
plt.show()
```

Because the data in the salary tend to have more values in same range and only a few values outside that range, we should not fill the missing values by using Mean.

So we use Median to fill the missing values.

```python
# using inplace parameter will fill the values in dataframe itself instead of returning a new copy of dataframe
df['salary'].fillna(df['salary'].median(), inplace=True)
```

```python
df.isnull().sum()
```

Replace median() with mean() / mode() whatever desired
