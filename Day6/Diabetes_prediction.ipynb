```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
# loading the diabetes dataset to a pandas DataFrame
df = pd.read_csv('diabetes.csv')
print(df.shape)
df.head()
```

```python
df.describe()
```

```python
df.value_counts('Outcome')
```

```python
df.groupby('Outcome').mean()
```

```python
# seperating data and labels

x = df.drop(columns = 'Outcome', axis=1)
y = df['Outcome']
```

```python
# Data Standardization
# by doing this, all the values that range from 0 to something will come into range 0 to 1

scaler = StandardScaler()
scaler.fit(x)
standardized_data = scaler.transform(x)

print(standardized_data)
```

```python
x = standardized_data
# putting standardized dtaa in x and labels ae already in y
```

```python
#splitting the training and testing data
# if stratify is not used then all diabetic will go to xtrain and non diabetic go to xtest
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2, stratify=y, random_state=2)
print(x.shape,xtrain.shape,xtest.shape)
```

ytrain has labels for xtrain data, and ytest has labels for xtest data

```python
#Training the model
classifier = svm.SVC(kernel='linear')
classifier.fit(xtrain,ytrain)
print(xtrain)
```

```python
# Model evaluation
# accuracy score on the training data
xtrain_prediction = classifier.predict(xtrain)
training_data_accuracy = accuracy_score(xtrain_prediction, ytrain)
print('Accuracy score of the training data : ', training_data_accuracy)
```

```python
xtest_prediction = classifier.predict(xtest)
testing_data_accuracy = accuracy_score(xtest_prediction, ytest)
print('testing_data_accuracy :', testing_data_accuracy)
```

```python
# Making a predictive system

inputdata = (5,116,74,0,0,25.6,0.201,30)  # Sample input

# Convert input data to a DataFrame with correct column names
input_df = pd.DataFrame([inputdata], columns=df.drop(columns='Outcome').columns)

# Standardize input data
std_data = scaler.transform(input_df)

# Print standardized data (for debugging)
print("Standardized Input:", std_data)

# Make prediction
prediction = classifier.predict(std_data)

# Output prediction result
print("Predicted Output:", prediction)
```
