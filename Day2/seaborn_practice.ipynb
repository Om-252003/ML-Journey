```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline
```

Built-in datasets from Seaborn

```python
# Total Bill vs Tips Received Dataset
tips = sns.load_dataset('tips') # This loads dataset into pandas Dataframe
print(type(tips))

tips.head()
```

```python
# Setting theme for plots globally
sns.set_theme()
```

```python

# data visualization 

sns.relplot(data=tips,
            x='total_bill',
            y='tip',
            col='time',
            hue='smoker',
            style='smoker',
            size='size'
           )
plt.show()
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python

sns.scatterplot(data=iris,
                x='sepal_length',
                y='petal_length',
                hue='species'
               )
plt.show()
```

```python


sns.scatterplot(data=iris,
                x='sepal_length',
                y='petal_width',
                hue='species'
               )
plt.show()
```

```python
titanic = sns.load_dataset('titanic')
print(titanic.shape)
titanic.head()
```

```python
sns.countplot(data=titanic, x='class')
plt.show()
```

```python
sns.countplot(data=titanic,x='survived')
plt.show()
```

```python
sns.barplot(data=titanic,
            x='sex',
            y='survived',
            hue='class',
            )
plt.show()
```
